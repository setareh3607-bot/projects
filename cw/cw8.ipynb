هوش مصنوعی:گسترده ترین حوزه است و هدف آن ساخت سیستم هایی است که کارهایی مانند استدلال حل مساله درک زبان و تصمیم گیری 
را انجام می دهد
یادگیری ماشین زیر مجموعه ای از هوش مصنوعی هست که در آن سیستم به جای دریافت قواعد صریح برای همه حالت ها الگوها را از داده ها یاد میگیردو از آن برای پیش بینی یا تصمیم گیری استفاده میکند
شبکه های عصبی :خانواده هایی از مدل های یادگیری ماشین هستند که با الهام  ساده شده از ساختار مغز طراحی شده اند این مدل ها ازواحدهایی محاسباتی به نام نورون تشکیل میشوند که معمولا در لایه های ورودی پنهان خروجی قرار میگیرند
یادگیری عمیق:زیرمجموعه‌ای از یادگیری ماشین است که عمدتاً از شبکه‌های عصبی با لایه‌های متعدد استفاده می‌کند. این لایه‌ها امکان یادگیری نمایش‌های پیچیده‌تر و سلسله‌مراتبی از داده را فراهم می‌کنند.
